```python
#This code fills missing values in the Review Rating column by replacing them
#with the median rating of
# the same category, making the dataset more complete while keeping existing
# ratings unchanged.

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#This code cleans and standardizes column names in your dataset by converting them
 #to lowercase, replacing spaces with underscores, and shortening one column name.
 #Finally, `df.columns` displays all the updated column names. This makes column
 #names more consistent and easier to use in Python.
 # and it end rename the column names

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
#create a column age_group
labels = ['young adult','adult','middle aged','senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
df[['age','age_group']].head(20)
```

```python
#create a coloumn purchase_frequency-days
# to easily analyze data to convert main data from texxt to numbers for customer behaviors super useful for our analysis

frequency_mapping={
    'Fortnightly': 14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    "bi Weekly":7,
    'Annually':365,
    'Every 3 Months':90,

}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
